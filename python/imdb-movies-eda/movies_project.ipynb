Q3. How do we clean the duration column?

```python
df["duration"]
```

```python
df["duration"] = df["duration"].str.replace("min","") # This operation is no longer needed as 'duration' is already an integer column.
```

```python
 df["duration"] = df["duration"].str.replace(",","") # This operation is no longer needed as 'duration' is already an integer column.
```

```python
df["duration"] = df["duration"].astype(int)
```

Q4. How do we clean the `gross` column and `votes` column?

```python
 df["votes"] = df["votes"].str.replace(",","") # This operation is no longer needed as 'votes' is already an integer column.
```

```python
df["votes"] = df["votes"].astype(int)
```

```python
 df['gross_income']= df['gross_income'].str.replace(",","") # This operation is no longer needed as 'gross_income' is already a float column.
```

```python
df['gross_income'] = df['gross_income'].str.replace("$","") # This operation is no longer needed as 'gross_income' is already a float column.
```

```python
df['gross_income'] = df['gross_income'].str.replace("M","") # This operation is no longer needed as 'gross_income' is already a float column.
```

```python
df['gross_income'] = df['gross_income'].astype(float)
```
